## In the following program we cluster the dataset n.1 with K-means.
From the plot of dataset n.1 we see 3 separated clusters. Thus k=3.

```python
from sklearn.cluster import KMeans

np.random.seed(5)

k=3
kmeans1 = KMeans(n_clusters=k, random_state=0).fit(dataset1)

i=0
nrighe_dataset1=dataset1.shape[0]

for i in range(nrighe_dataset1):
    print("Example n."+str(i)+"=("+str(dataset1[i,0])+","+str(dataset1[i,1])+")")
    print("in cluster n."+str(kmeans1.labels_[i]))
```

We plot the clusters with a pandas data frame:

```python
clustering_results_dataset1 = []

for i in range(nrighe_dataset1):

    clustering_results_dataset1.append({
        'Instance': i,
        feature_names1[0]: dataset1[i,0],
        feature_names1[1]: dataset1[i,1],
        'Cluster n.': kmeans1.labels_[i]
    })

results_df1 = pd.DataFrame(clustering_results_dataset1 )
print(results_df1.to_string(index=False, float_format='%.1f'))
```

In the following program we plot the clusters

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(8,8))

ax = fig.add_subplot(111)
fig.subplots_adjust(top=1)
ax.set_title('Clustered points in dataset n. 1')

ax.set_xlabel('x')
ax.set_ylabel('y')

# set the list of colors to be selected when plotting the different clusters
color=['b','g','r','c','m','y','k','w']

#plot the dataset
for clu in range(k):
    # collect the sequence of cooordinates of the points in each given cluster (determined by clu)
    data_list_x = [dataset1[i,0] for i in range(nrighe_dataset1) if kmeans1.labels_[i]==clu]
    data_list_y = [dataset1[i,1] for i in range(nrighe_dataset1) if kmeans1.labels_[i]==clu]
    plt.scatter(data_list_x, data_list_y, s=2, edgecolors='none', c=color[clu], alpha=0.5)

plt.show()
```

In the following cells I propose you to write a program that computes with a statistical measure of your choice a quantitative evaluation of clusters in the three datasets.

**Note:**
It is advisable to execute K-means a certain number of times (let us try 10 times) and then select the clustering solution that gives the best value of the evaluation measure.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import davies_bouldin_score

def best_dbi(data, k, tries=10, seed=0):
    best = None
    for i in range(tries):
        km = KMeans(n_clusters=k, random_state=seed + i, n_init=10).fit(data)
        dbi = davies_bouldin_score(data, km.labels_)
        if best is None or dbi < best:
            best = dbi
    return best

dbi_1 = best_dbi(dataset1, k=3)
dbi_2 = best_dbi(dataset2, k=k2)
dbi_3 = best_dbi(dataset3, k=k3)

print("DBI best dataset1:", round(dbi_1, 3))
print("DBI best dataset2:", round(dbi_2, 3))
print("DBI best dataset3:", round(dbi_3, 3))

#La DBI misura quanto i cluster sono compatti e separati: piu' e' bassa, 
#migliore è la clusterizzazione.
```

Plot in the cell below the quantitative measure of your choice (used above) with respect to an increasing value of k (the number of clusters) so that the best value of k can be selected.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import davies_bouldin_score

def dbi_curve(data, k_values, seed=0):
    scores = []
    for k in k_values:
        km = KMeans(n_clusters=k, random_state=seed, n_init=10).fit(data)
        scores.append(davies_bouldin_score(data, km.labels_))
    return scores

k_values = list(range(2, 9))
s1 = dbi_curve(dataset1, k_values, seed=0)
s2 = dbi_curve(dataset2, k_values, seed=1)
s3 = dbi_curve(dataset3, k_values, seed=2)

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(k_values, s1, marker='o', label='dataset1', linewidth=2)
ax.plot(k_values, s2, marker='o', label='dataset2', linewidth=2)
ax.plot(k_values, s3, marker='o', label='dataset3', linewidth=2)
ax.set_xlabel('k', fontsize=12)
ax.set_ylabel('Davies-Bouldin Index', fontsize=12)
ax.set_title('Davies-Bouldin Index vs Number of Clusters', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=11)
plt.tight_layout()
plt.show()

#Nel grafico il k migliore e' dove la DBI raggiunge il valore minimo.
```

In the following cell I propose you to:

1.   Set *Minpts* to a number (say **10**).
2.   Compute the **reachability distance** of the **10-th nearest neighbour ** for each data-point.
3. Sort the set of reachability distances you obtained in an increasing way.
4. Plot the sorted reachability distances
5. Find the **elbow** of the diagram => it gives the eps value combined with Minpts=10.
6. Try this combined pair of parameters on the dataset you chose, with DBSCAN.

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

#ancora, scaliamo i dati per DBSCAN, che è sensibile alle scale delle feature 
# come buona pratica, non è obbligatorio, ma aiuta a ottenere risultati migliori
data2_scaled = StandardScaler().fit_transform(dataset2)

minpts = 10
nbrs = NearestNeighbors(n_neighbors=minpts).fit(data2_scaled)
distances, _ = nbrs.kneighbors(data2_scaled)
k_dist = np.sort(distances[:, -1])

#Trova l'elbow point con il metodo della distanza dalla linea
#Questo metodo trova il punto piu' distante dalla linea retta che connette 
#il primo e l'ultimo punto della curva (cioe' dove la curva "piega" di piu')
def trova_elbow_point(y_values, start=0):
    #Trova il punto della curva piu' distante dalla linea retta
    n = len(y_values)
    #Definisce la linea retta tra primo e ultimo punto
    p1 = np.array([start, y_values[0]])
    p2 = np.array([start + n - 1, y_values[-1]])
    
    #Calcola la distanza di ogni punto dalla linea
    distanze = []
    line_vec = p2 - p1  #vettore della linea
    line_len = np.linalg.norm(line_vec)  #lunghezza della linea
    
    for i in range(n):
        punto = np.array([start + i, y_values[i]])
        point_vec = punto - p1
        #Formula della distanza punto-linea in 2D (prodotto vettoriale come scalare)
        cross_prod = line_vec[0] * point_vec[1] - line_vec[1] * point_vec[0]
        dist = abs(cross_prod) / line_len
        distanze.append(dist)
    
    return np.argmax(distanze)

#Applica il metodo partendo dal punto 1000 per evitare rumore iniziale
start_idx = 1000
elbow_idx_relativo = trova_elbow_point(k_dist[start_idx:], start_idx)
elbow_idx = elbow_idx_relativo + start_idx
eps_optimal = k_dist[elbow_idx]

#Visualizza il grafico con l'elbow point individuato
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

#Subplot 1: K-distance plot completo con elbow
axes[0].plot(k_dist, linewidth=1.5, color='steelblue', label='k-distance curve')
axes[0].axvline(x=elbow_idx, color='red', linestyle='--', linewidth=2, label=f'Elbow point (idx={elbow_idx})')
axes[0].scatter([elbow_idx], [eps_optimal], color='red', s=100, zorder=5, label=f'eps={eps_optimal:.4f}')
axes[0].set_xlabel('Data Points (sorted)', fontsize=12)
axes[0].set_ylabel('k-Distance (k=10)', fontsize=12)
axes[0].set_title('K-distance Graph for DBSCAN Parameter Selection (minpts=10)', fontsize=13, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)

#Subplot 2: Zoom sulla zona dell'elbow
zoom_range = 1500
zoom_start = max(0, elbow_idx - zoom_range)
zoom_end = min(len(k_dist), elbow_idx + zoom_range)
axes[1].plot(range(zoom_start, zoom_end), k_dist[zoom_start:zoom_end], linewidth=2, color='steelblue')
axes[1].axvline(x=elbow_idx, color='red', linestyle='--', linewidth=2)
axes[1].scatter([elbow_idx], [eps_optimal], color='red', s=150, zorder=5)
axes[1].set_xlabel('Data Points (sorted)', fontsize=12)
axes[1].set_ylabel('k-Distance', fontsize=12)
axes[1].set_title('Zoom sulla zona dell\'elbow', fontsize=13, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Elbow point trovato all'indice: {elbow_idx}")
print(f"Valore eps raccomandato: {eps_optimal:.4f}")
```

```python
from sklearn.cluster import DBSCAN

#Applica DBSCAN con l'eps ottimale trovato dall'analisi della k-distance
dbscan = DBSCAN(eps=eps_optimal, min_samples=minpts).fit(data2_scaled)
labels_final = dbscan.labels_

#Conta i cluster, qui -1 rappresenta i rumori, quindi li escludiamo dal conteggio dei cluster reali
unique_labels = set(labels_final)
n_clusters = len(unique_labels) - (1 if -1 in unique_labels else 0)

#Visualizza i risultati
fig, ax = plt.subplots(figsize=(10, 8))

#Risultati DBSCAN nello spazio 2D originale
#la maschera serve per selezionare solo i punti che sono stati assegnati a un cluster (labels_final != -1)
cluster_mask = labels_final != -1

if np.any(cluster_mask):
    ax.scatter(dataset2[cluster_mask, 0], dataset2[cluster_mask, 1], 
                        c=labels_final[cluster_mask], cmap='tab20', s=20, 
                        edgecolors='black', linewidth=0.5, label='Clusters')

ax.set_xlabel(feature_names2[0], fontsize=12)
ax.set_ylabel(feature_names2[1], fontsize=12)
ax.set_title('DBSCAN Clustering Results (eps={:.4f}, minpts={})'.format(eps_optimal, minpts), fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"eps: {eps_optimal:.4f}, minpts: {minpts}")
print(f"Clusters found: {n_clusters}")
print(f"Noise points: {(labels_final == -1).sum()}")
```
